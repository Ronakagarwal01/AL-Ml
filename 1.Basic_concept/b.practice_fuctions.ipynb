```python
# c 

info = [
    ('alice','math'),
    ('bob','science'),
    ('alice','science'),
    ('charlie','math'),
    ('bob','math'),
    ('alice','English'),
    ('charlie','English')

]


dict = {}

for name , course in info:
    if (dict.get(name) == None):
        dict.update({name:set()})
        dict[name].add(course)

    else:
        dict[name].add(course)

print(dict)
```
