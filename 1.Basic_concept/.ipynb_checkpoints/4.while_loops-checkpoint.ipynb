```python
# finite loop using while 

count = 1

while (count <=5):
    print("hello world",count)
    count += 1

print("after loop count value is ",count)
```

```python
# 1. print no. 1 to 5 

i = 1 

while (i <=  5):
    print(i)
    i += 1
```

```python
# 2. reverse no. 5 to 1 

i = 5 

while (i >= 1 ):
    print(i)
    i -= 1
```

```python
# 3. print a multiple table of n 

n = int(input("Enter a number: "))

i = 0

while (i <= 10):

    print(n * (i+1))

    i=i+1

   
```

```python
# 4. print vowel count of a given string 

word = "artificial intellegence"

count = 0

for ch in word:
    if (ch =='a' or ch =='e' or ch =='i' or ch == 'o' or ch == 'u'):
        count = count + 1

print("ans = ",count)
    
```
