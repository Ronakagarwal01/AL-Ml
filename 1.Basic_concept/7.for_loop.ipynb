```python
# 1. count the i word in ai

word = " artificial intelligence"

# count the no. of i's => 5

count = 0

for ch in word:
    if (ch == 'i'):
        count += 1

print("count of i = " ,count)

```

```python
# 2. print sun of first 'n' natural numbers 

n = int(input("Enter a number: "))

sum = 0

for i in range(1,n+1):

    sum = sum + i
print("sum = " , sum)

```
