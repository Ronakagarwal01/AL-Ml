## dictionary methods

```python
info = {

    'name': 'ronak',
    'cgpa':9.2,
    'subjects':['maths','science'],
    3.14:'pi'
}

dict_keys = info.keys()
print(dict_keys)
```

```python
dict_values = info.values()

dict_values
```

```python
info = {

    'name': 'ronak',
    'cgpa':9.2,
    'subjects':['maths','science'],
    3.14:'pi'
}


info.update({
    'city':'delhi'
})

print(info)
```

```python
print(info.get('cgpa23'))  ## if there was no any key pair in dictnory get function pass none , not error
```
