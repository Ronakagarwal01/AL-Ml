```python
# json string ko python object me convert karna using loads

import json

json_str = '{"name ":"ronak", "isTeacher":null}'

py_obj = json.loads(json_str)

print(type(json_str))

print(type(py_obj) , py_obj)
```

```python
# python object ko json string me convert karna using dumps 

import json

py_obj = {
    "name":"Ronak",
    "isTeacher":None
}

json_str = json.dumps(py_obj)

print(type(json_str) , json_str)
```

```python
# reading json data from a file using load

import json 

with open ("27.data.json" , "r") as f:
    py_obj = json.load(f)
    print(type(py_obj) , py_obj)
```

```python
# converting the data from python object to json formet using dump in a file 

import json 

data = {
    "name ":"Ronak",
    "age" : 21,
    "isTeacher" :True

}

with open("27.data.json" , "w") as f:
    json.dump(data , f , indent = 4 ,sort_keys = True) 
```
