```python
# basis code for print square fore 0 to 6 in a list 

sqr = []

for i in range (0,6):
    sqr.append(i*i)

print(sqr)
```

## using List Comprehension

```python
# example
sq = [i*i for i in range(6)]
print(sq)
```

```python
# only odd no. square 

square = [ i*i for i in range(6) if i%2 != 0]

square
```

```python
# we have already list [-2,-4,3,5,2,-1] and the negative no. are replaced by o and we get [0,0,3,5,2,0]

nums = [-2,-4,3,5,2,-1]

nums = [ 0 if val < 0 else val for val in nums ]
print(nums)
```

```python
# program for doing all words in upper case 

words = ['hello', 'python' , 'apnacollege']

words = [valu.upper() for valu in words]

print(words)
```
