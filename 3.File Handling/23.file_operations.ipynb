## Modes in file operations

```python
# write append 


f = open("sample.txt","a")

f.write("\nnew text have been added  \n check the paragraph")

f.close()
```

```python
# create new file and open for writing

f = open("sample2.txt",'x')  

f.write('hello everyone \n what are you all doing !!')

f.close()
```

```python
# r+ mode


f = open("sample.txt",'r+')  

f.write("123")
print(f.read())

f.close()
```

```python
# a+ mode


f = open("sample.txt",'a+')  

f.write("123")
print(f.read())

f.close()
```

```python
# w+ mode

f = open("sample.txt",'w+')  

f.write("123")
print(f.read())

f.close()
```

# with keyword


```python
with open("sample.txt",'r') as f:
    data = f.read()
    print(data)
    print(len(data))
```
