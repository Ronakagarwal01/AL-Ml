```python
import numpy as np
import time 
```

```python
# Execution performance
size =1_000_000


#python list 
py_list = list(range(size)) # calculate the square in this range 
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"pyhton list time = {end-start} secounds ")


# numpy arrays
np_arr = np.array(py_list)
start = time.time()

#vectorization
sq_array = np_arr ** 2

end = time.time()

print(f"numpy list time = {end-start} secounds ")
```
