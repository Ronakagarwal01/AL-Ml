```python
# Standerd Normalization = (arr - mean) / standerd devition


```

```python
import numpy as np
```

```python
arr = np.array([[1,2],[3,4]])

mean =  np.mean(arr)
std_dev = np.std(arr)

normalized_arr = (arr - mean)/ std_dev

print(normalized_arr )
print(std_dev)
```
