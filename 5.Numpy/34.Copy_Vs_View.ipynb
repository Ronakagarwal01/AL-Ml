```python
import numpy as np
```

```python
# in list when we can do slicing then the slicing create a new space in the memory 
nums = [1,2,3,4,5]
sub_list = nums[1:3]
print(sub_list)
sub_list[0] = 200

print(sub_list)
print(nums)


```

```python
# but in array slicing dont take extra place in memory it shows the original array

arr = np.array([1,2,3,4,5])
sub_array = arr[1:3]
print(sub_array)
sub_array[0] = 200

print(sub_array)
print(arr)
```
