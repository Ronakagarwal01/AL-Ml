```python
# Data type
```

```python
import numpy as np

arr = np.array([1,2,3,4,5])

print(arr, arr.dtype)
```

```python
arr1 = np.array([1.0,2,3,4,5])

print(arr1, arr1.dtype)
```

```python
arr2 = np.array([1.0,2,3,4,5, "prime"])

print(arr2, arr2.dtype)
```

```python
# complex no. 
arr3 = np.array([3+5j])
arr4 = np.array([2+3j])

print(arr3 + arr4 ,arr3.dtype)
print(arr3 - arr4)
```
