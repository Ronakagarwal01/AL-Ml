```python
#important mathematical Function

import numpy as np
```

# Aggregation fnx

```python
arr = np.array([1,2,3,4,5])


# Aggregation fnx
print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr))  # it shows the index number of  minimum value 
print(np.argmax(arr))  # it shows the index no. of maximum value 


print(np.mean(arr))
print(np.std(arr))
print(np.median(arr))
print(np.var(arr))
```

# Power fnx

```python
# Power fnx

arr = np.array([1,2,3,4,5])

print(np.square(arr))
print(np.sqrt(arr))
print(np.power(arr , 3))
```

# log fnx

```python
# log fnx

arr = np.array([1,2,3,4,5])

print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))


print(np.exp(arr))   # for exponctial fnx 

```

# Rounding function

```python
# Rounding function 
print(np.round(3.14))  # Round off

print(np.ceil(3.14))  # Round Up

print(np.floor(3.14))  # Round down
print(np.trunc(3.6768614))  # Remove value after decimal 
```

```python
arr = np.array([1,2,3,4,5,5,5,6,6,7,7,7,1,-2,-4,-3,-5])

print(np.unique(arr))   # gives all the unique values 
print(np.sort(arr))     # sort the array like sets 
print(np.abs(arr))      # convert negative values to positive 
```
