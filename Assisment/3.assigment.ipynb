```python
# 1. Ask the user for a string and check whether it is a palindrome or not.

str = input("Enter a word ")


if str[::-1] == str:
    print(f" {str} is a palindrom")

else:
    print(f"{str} is not a palindrom")
```

```python
# 2. Given a list of integers compute the average of all numbers in the list

list = [5,6,7,8,9,5,4,15564,7]

sum = 0
for i in list:
    sum = sum + i

average = sum / len(list)

average
```

```python
# 4. Given a tuple of integers , create:

# a. a tuple of all even numbers
# b. a tuple of all oddd numbers


t = (1,2,3,4,5,6,7,8,9)


even = ()
odd = ()

for i in t:
    if i % 2 == 0:
        even = even + (i,)

    else:
        odd = odd + (i,) 

print(even)
print(odd)
```
