```python
# 2. Take two numbers as input from the user and print their sum ,difference , product and quotient.


a = int(input('enter a number '))
b = int(input('enter a number '))

sum =  a + b
difference = a - b
product = a*b
quotient = a/b

print("sum is ", sum)
print("difference is ", difference)
print("product is ", product)
print("quotient is ", quotient)
```

```python
# 3.  Ask the user to enter two integers and one float. Convert them all to floats and print their average.

a = int(input("enter first integer"))
b = int(input("enter second integer"))
c = float(input("enter float value"))

average = (float(a) + float(b) + c)/ 3

print("the average of three value is ",average)
```

```python
# 7. Ask the user for a temperature in Celsius (string input). Convert it to float ,
# then calculate and print temperature in Fahrenheit.

# conversion formula : fahrenheit = (celsius * 9/5) + 32

celsius = float(input("enter temperature in celsius"))

fahrenheitTemp = (celsius * 9/5) + 32
print("temperature in fahrenheit is ",fahrenheitTemp)
```

```python
# 9. ask the user for : principle (p) ,Rate (R), Time (T). convert all to float and compute simple intrest:

# Simple intrest formula : SI = (P * R * T)/100


principle = float(input("Enter the principle amount "))
rate = float(input("Enter the value of interest rate "))
time = float(input("Enter the time "))

SI = (principle * rate  * time )/100

print("The simple intrest is ",SI)
```

```python
# 10. Take a decimal number as input (like 45.79) and output its:

# a. integer part - 45 
# b. fractional part - 0.79



decimal_number = float(input("Enter a decimal number: "))
integer_part = int(decimal_number)
fractional_part = round(decimal_number - integer_part, 2)

print("The integer part is", integer_part)
print("The fractional part is", fractional_part)

```
