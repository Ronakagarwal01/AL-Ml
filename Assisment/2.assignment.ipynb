```python
# write a function that takes two  integer a and b and prints all even number between them 

def even(a,b):

    # ensure a is less than b 

    if a>b:
        a,b = b,a

    for i in range(a,b+1):
        if (i%2==0):
            print(i ,end = " ")

# input
a = int(input())
b = int(input())

even(a,b)
```

```python
# 7. design a program to continously input a number n from a user and print if it is positive or negative untill the user enters 'quit'


while True:
    n = input("Enter a number (or 'quit to stop):")

    if n == 'quit':
        break

    n = int(n)

    if n > 0:
        print('positive')

    elif n < 0:
        print("negative")

    else:
        print('zero')
```

```python
# 8. lets create a simple calacuator that perform operations . create a function calculator (a,b,oprator)


def calculator (a,b,operator):

    if operator == '+':
        sum = a+b
        return sum 
        

    elif operator == '-':
        sub = a-b
        return sub
        

    elif operator == '*':
        mult = a*b
        return mult

    elif operator == '/':
        div = a/b
        return div

    else:
        return 'invalid input'


    


a = int(input("Enter a first number"))
b = int(input("Enter secound number "))
operator = input("Enter a operator ")

result = calculator(a,b,operator)
result
```

```python
# 9. write a function is_prime(n) that returns True if n is prime number and false otherwise, using loop.


def is_prime(n):
    if n <= 1:
        return False
    
    for i in range(2,n):
        if n % i == 0:
            return False

        

    return True

n = int(input("Enter a number "))

is_prime(n)
```
