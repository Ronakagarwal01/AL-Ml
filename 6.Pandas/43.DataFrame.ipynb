```python
import pandas as pd
```

```python
# 1st way to create a datafram 
info = {
    "Nmae":['adam','bob','eve'],
    "Age":[23,24,25],
    'cgpa':[7.8,8.9,9.5]
}

df = pd.DataFrame(info)

print(df)
print(df.index)
print(df.columns)
```

```python
# secound way to create a dataframe using lists of list 

df = pd.DataFrame([['adam',29,8.7],['jarry',21,8.6],['Rohit',25,6.9]] , columns = ['name','age','cgpa'])

df
```

```python
# third way to create a dataframe is using numpy array

import numpy as np

np_arr =np.array([[1,2,3],[4,5,6]])

df = pd.DataFrame(np_arr , columns = ["A","B","C"])
df
```
