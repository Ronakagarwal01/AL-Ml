# BeautifulSoup Activity

### From Quotes ,extract all quotes & authors from all pages with the life tag.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Fetch data and store it 

```python
#Fetch data and store it 


page_count = 1

while True:
    url = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(url)

    


    soup = BeautifulSoup(res.text,'lxml')
    quotes = soup.select('div.quote')


    if not quotes:
        print('No valid pages anymore ..')
        break

    with open(f"scraped_data/quotes{page_count}.html","w" ,encoding="utf-8") as f:
        f.write(res.text)

        print(f'we have sucessfully downloaded data from {page_count} ')

    page_count = page_count + 1
        
        
```

## Extract useful info

```python
all_soups = []

for i in range(1,11):

    with open(f'scraped_data/quotes{i}.html', 'r', encoding='utf-8') as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, 'lxml')

    all_soups.append(soup)
```

```python
all_quotes = []

for soup in all_soups:
    quotes = soup.select('div.quote')
    all_quotes.extend(quotes)
```

```python
life_quotes = []

for q in all_quotes:
    all_tags = []

    for tag in q.select('.tags .tag'):
        all_tags.append(tag.get_text())

    if 'life' in all_tags:
        text = q.select_one('span.text').get_text()
        author = q.select_one('small.author').get_text()

        life_quotes.append([text, author])

life_quotes
```

```python
df = pd.DataFrame(life_quotes,columns = ['Quotes','Author'])
```

```python
df.to_csv('cleaned_data/data2.csv')
```
