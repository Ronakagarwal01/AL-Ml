```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "country":['USA','USA','India','India'],
    "year":[2020,2021,2020,2021],
    "sales": [100,120,90,110],
    'profit': [20,25,18,22],
    'tax':[10,20,30,40]
})
```

```python
df
```

```python

# using this we can reduce the columns 

melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales","profit","tax"],
    var_name = "metrics",
    value_name= "value"   
)

melted_df
```
