```python
import pandas as pd 
```

```python
# Concatenation

df1 = pd.DataFrame({
    'id': [1,2,3],
    "name": ['adam','eiv','bob']
})


df2 = pd.DataFrame({
    'id': [4,5,6],
    "name": ['charlie','fester','gilian']
})

```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2]) # Row based concationation
pd.concat([df1,df2], ignore_index = True) # Row based concationation
```
