```python
import pandas as pd
```

```python
df = pd.read_csv('raw_data.csv')
df
```

# apply fnx

```python
# feature engineering 

df2 = df.copy()

df2['tax'] = df2['income'].apply(lambda x:'20%' if x>= 60000 else "10%")

df2
```

# Map fnx

```python
# suppose we want to convert Male => M , Female => F , Unknown => U

gender_map = { 'Male' : "M" , "Female":'F',"Unknown":'U' }
df2['gender'] = df2['gender'].map(gender_map)
df2
```

# Assign fnx

```python
df2=  df2.assign(new_income = df2['income'] * 1.1 )
df2
```

```python
df2
```

# Replace Fnx

```python
df2['country'] = df2['country'].replace('USA',"United State")
df2
```

```python
# some more fnx 
```

```python
# change column names 

df2.columns = ['Id',"Name","Age","Country","Gender","Income","Tax","New_Income"]
df2
```

# Rename fnx

```python
df2.rename(columns={'Income':'Salary'})
```

# Sort Values

```python
# we want to sort the data on the basis of income 

df2['Age'] = df2['Age'].fillna(50)

df2.sort_values('Income')
df2.sort_values(['Income'],ascending = False)
sorted_df = df2.sort_values(['Income' , 'Age'])
sorted_df
```
