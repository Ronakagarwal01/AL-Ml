# Compare different dataset on bar


```python
oscar_movies = [
    "The Dark Knight",
    "The hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] #in $M

non_oscar_movies = ['Slumdog Millinaire',"Avatar","Inception","Hugo","Lincoin"]
non_oscar_revenue = [378,2788,829,185,275] #in M$

years = [2008, 2009, 2010, 2011, 2012]

x= np.arange(len(years))
width = 0.4

plt.bar(x-width/2 , oscar_revenue, width, label = "Oscar Movies")
plt.bar(x+width/2 , non_oscar_revenue ,width , label = "Non_Oscar Movies")


plt.title('Oscar vs non oscsar revenue')
plt.xlabel("Year")
plt.ylabel('Revenue ($M)')
plt.legend()

plt.xticks(x,years)
```
