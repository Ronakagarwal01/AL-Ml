```python
import matplotlib.pyplot as plt
```

# Pie Charts

## Prefer them only for few categories

```python
expence = ['salaries',"Rent","Marketing","R&D","Miscellaneous"]
amount =[500,150,120,100,50]


plt.pie(amount , labels= expence)
```

# Customization

```python
expence = ['salaries',"Rent","Marketing","R&D","Miscellaneous"]
amount =[500,150,120,100,50]


plt.pie(amount , labels= expence , autopct ="%1.1f%%")
plt.show()
```
