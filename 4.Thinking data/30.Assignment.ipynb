```python
import json 
```

```python
# Load the data 

def load_data(filename):
    with open (filename,"r") as f:
        data = json.load(f)

    return data
        
```

```python
data = load_data('store_data.json')

print(data)
print(type(data))
```

### Assignment 1 

```python
# cleaning and stracturing our data
```

```python
# clean & stracture the data  
def clean_data(data):  
    text_to_numbers ={ "one":1 , "two":2 , "three":3 , "four":4 , "five":5}
    
    cleaned_data = []
    unique_users = set()

    
    for user in data:
        # clean the rating part 
        original_rating =str(user['rating']).strip().lower() # strip()  =  "  5  "  => "5"        
        if(original_rating in text_to_numbers):
            original_rating = text_to_numbers[original_rating]
        user['rating'] = original_rating


        # handle our missing value
        original_age = user.get("age")
        if(original_age == None):
            user['age'] = None

        # deduplication
        if(user['name'].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
    
```

```python
data = clean_data(data)
data
```

### Assignment 2

```python
#1. Get meaningful insights from the data 
def get_insights(data):

    # average rating of the product
    total_rating = 0
    for user in data:
        total_rating = total_rating + float(user['rating'])
    print(f"average_rating = {total_rating / len(data)}")


    # percentage of user with poore rating 

    poor_rating = 0 
    for user in data:
        if(float(user['rating']) < 3):
            poor_rating += 1

    print(f"% of user with the poor rating = {poor_rating / len(data) *100} % " )
        
```

```python
get_insights(data)
```

### Assignment 3

```python
# build a product recommendation feature (Rule based , not ML)

# 1. if user rating >= 4 , recommend same brand products.
# 2. is user rating < 4 , recommand different brand products
```

```python
def get_recommendation(data):
    recommendation = []

    for user in data:

        current_recommendation = {}
        current_recommendation['name'] = user['name']
        if float(user['rating']) >= 4:
            current_recommendation['brand'] = 'Apple'
        else:
            current_recommendation['brand'] = 'samsung'

        recommendation.append(current_recommendation)
    
    return recommendation
```

```python
get_recommendation(data)
```
