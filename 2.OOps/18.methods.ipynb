## 1.instance method


```python
# instance method


# it can access class as well as  instance attributes
class laptop:
    storage_type = 'ssd'

    def __init__(self,RAM,storage):
        self.RAM = RAM
        self.storage = storage

    def get_info(self): # instance method
        print(f' laptop has {self.RAM} RAM and {self.storage} {self.storage_type}')

l1 = laptop('16gb','512gb')
l2 = laptop('8gb','256gb')


l1.get_info()
```

## 2. class method

```python
# class only acces class method not instancs method 



class laptop:
    storage_type = 'ssd'

    def __init__(self,RAM,storage):
        self.RAM = RAM
        self.storage = storage


    @classmethod
    def get_storage_type(cls):
        print(f' storage type = {cls.storage_type}')

    def get_info(self): # instance method
        print(f' laptop has {self.RAM} RAM and {self.storage} {self.storage_type}')

l1 = laptop('16gb','512gb')


laptop.get_storage_type()
l1.get_storage_type()
```

## 3. Static Method

```python
class laptop:
    storage_type = 'ssd'

    def __init__(self,RAM,storage):
        self.RAM = RAM
        self.storage = storage


    @classmethod
    def get_storage_type(cls):
        print(f' storage type = {cls.storage_type}')

    def get_info(self): # instance method
        print(f' laptop has {self.RAM} RAM and {self.storage} {self.storage_type}')



    # static
    @staticmethod
    def cal_discount(prize , discount):
        final_prize = prize  - (discount * prize / 100)
        print(f'discount prize = {final_prize}')




li = laptop('16gb','512gb')

li.cal_discount(40_000 , 10)



## static 

# fnx = (prize , discount)  => final prize
```
