# Inheritance

### Reusing attributes and methods from a parent (Base) class

```python
# 1. 

class Employee:
    start_time = '10am'
    end_time = '6pm'


    def change_time(self ,new_end_time):
        self.end_time = new_end_time

class Teacher(Employee):    # here we can use Employee class so it can perform inheritance 
    def __init__(self,subject):
        self.subject = subject


class AdminStaff(Employee):
    def __init__(self ,role ):
        self.role = role 


t1 = Teacher('Math')
t1.change_time('5pm')

print(t1.subject , t1.start_time , t1.end_time)


staff1 = AdminStaff('manager')

print(staff1.role , staff1.start_time , staff1.end_time)

```

```python
# 2. multilevel inheritance

class Employee:
    start_time = '10am'
    end_time = '6pm'

class AdminStaff(Employee):
    def __init__(self ,role ):
        self.role = role 

class Accountant(AdminStaff):
    def __init__(self ,salery , role ):
            super().__init__(role)
            self.salery = salery 


acc1 = Accountant(25_000 , 'CA')


print(acc1.salery , acc1.role , acc1.start_time , acc1.end_time )
```

```python
# 3. multiple inheritance 

class Teacher:
    def __init__(self , salery):
        self.salery = salery

class Student:
    def __init__(self, cgpa):
        self.cgpa = cgpa

class TA(Teacher , Student):

    def __init__(self,salery , cgpa , name):
        super().__init__(salery)
        Student.__init__(self ,cgpa)
        self.name = name 

ta1 = TA(150000 , 9.3 , "Ronak")

print(ta1.name , ta1.cgpa ,ta1.salery)
```
