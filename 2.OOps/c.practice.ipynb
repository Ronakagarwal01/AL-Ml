```python
# 3. 

class product:
    
    def __init__(self , name , prize):
        self.name = name 
        self.prize = prize 

    def get_info(self):
        print(f'prize of {self.name} is Rs. {self.prize}' )


    @staticmethod
    def calc_discount(prize , discount):
        print(f'discounted prize = {prize - (prize * discount /100)}')



p1 = product('phone' , 10_000)
p2 = product ('laptop', 40_000)
p3 = product('pen',10)

p1.calc_discount(45_000,10)
```
