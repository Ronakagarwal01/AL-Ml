# Abstruction

### hiding internal details & showing only essential features

```python
from abc import ABC ,abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Lion(Animal):
    def make_sound(self):
        print("Rore!!")


class Cow(Animal):
    def make_sound(self):
        print("Coo!!!!")


lion = Lion()
lion.make_sound()


cow = Cow()
cow.make_sound()


```
