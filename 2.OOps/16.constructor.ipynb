```python
class student:
    def __init__(self,name,cgpa):
        self.name = name
        self.cgpa = cgpa
      

stud1 = student('Rahul', 7.5)
stud2 = student('urvashi',6.3)
stud3 = student('ronak', 7.23)

print(stud1.name, stud1.cgpa)
print(stud2.name)
print(stud3.name)
```

```python
class student:
    def __init__(self,name,cgpa):
        self.name = name
        self.cgpa = cgpa

    def get_cgpa(self):
        return self.cgpa

stud1 = student('Rahul', 7.5)
stud2 = student('urvashi',6.3)
stud3 = student('ronak', 7.23)



print(f' {stud1.name} has cgpa = {stud1.get_cgpa()}')
```
