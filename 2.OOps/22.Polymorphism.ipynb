```python
# 1. Function overriding

class Employee:
    def get_desination(self):
        print('designation = Employeee')

class Teacher(Employee):
    def get_desination(self):
        print('designation  = Teacher ')


t1 = Teacher()
t1.get_desination()
```

```python
# 2, duck typing


class Teacher():
    def get_desination(self):
        print('designation  = Teacher ')

class Accoutant():
    def get_desination(self):
        print('designation  = accountant ')

t1 = Teacher()
t1.get_desination()


acc1 = Accoutant()
acc1.get_desination()
```
